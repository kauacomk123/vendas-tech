```python
import pandas as pd
import numpy as np

df_vendas = pd.read_csv("vendas_tech.csv")
display(df_vendas.head(10)) #imprime os 10 primeiros dados da base de dados
#display(df_vendas.tail(10)) #imprime os 10 ultimos
# df_analise = df_vendas
# df_analise["Data_Base"] = df_analise['Data_Base'].fillna('indisponivel') #adiciono na coluna Data_base indisponiveis nos valores vazios
df_analise = df_vendas.drop(columns=['Data_Base']) #retiro a coluna Data_Base
# df_analise = df_analise.dropna() #Remover linhas com valores vazios:

df_analise['Loja'] = df_analise['Loja'].fillna('Online') #adiciona 'online' nas celulas vazias na coluna loja 
df_analise['Data'] = pd.to_datetime(df_analise['Data'], format='%Y-%m-%d') #padronizaçao de data 
df_analise['Loja'] = df_analise['Loja'].str.strip().str.title() #padronizar os dados da coluna Loja deixando apenas a primeira letra de cada palavra em maiúscula
display(df_analise.head(10))
# O símbolo [:100] significa: do início até a linha 100
df_analise[:100].to_excel("vendas_tech_amostra.xlsx", index=False)
```

```python
df_analise['Loja'] = df_analise['Loja'].str.strip().str.title() #.str: acessa métodos de manipulação de texto;
#.strip(): remove espaços em branco no início e no fim do texto;
#.title(): coloca a primeira letra de cada palavra em maiúscula.
df_analise = df_analise.drop_duplicates()
df_analise = df_analise.drop_duplicates(subset=['ID_Pedido'], keep='first')

df_analise['forma_venda'] = np.where(
    df_analise['Loja'] == 'Online',
    'Online',
    'Presencial'
)

dic_regioes = {
    'São Paulo': 'Sudeste',
    'Rio De Janeiro': 'Sudeste',
    'Belo Horizonte': 'Sudeste',
    'Salvador': 'Nordeste',
    'Recife': 'Nordeste',
    'Curitiba': 'Sul',
    'Porto Alegre': 'Sul',
    'Online': 'Online'
}

df_analise['regiao'] = df_analise['Loja'].map(dic_regioes)
display(df_analise.head(10))

```

```python
df_filtrado = df_analise[
    (df_analise['Produto'] == 'Cabo HDMI') &
    (df_analise['regiao'] == 'Nordeste')
]

df_filtrado2 = df_analise[
    (df_analise['Produto'] == 'Mouse Gamer') |
    (df_analise['regiao'] == 'Nordeste')
]
display(df_filtrado.head(10))
display(df_filtrado2.head(10))
```

```python
df_analise['faturamento'] = df_analise['Qtd'] * df_analise['Preco_Unitario']

display(df_analise.head(10))
```
